# check what sites don't have coordinates

```python
import json
from pathlib import Path
from collections import defaultdict


def find_new_sites_missing_coordinates(sessions_folder: str) -> dict:
    """
    Scan all JSON input files in sessions folder.
    Find new sites (##new_cc_siteN##) with missing lat/lon.
    
    Args:
        sessions_folder: path to folder with JSON files (e.g., "./sessions/26_06")
    
    Returns:
        {
            "missing_coords": [
                {
                    "site": "##new_cc_site5##",
                    "scenario": "Preferred",
                    "year": "2030",
                    "file": "path/to/json",
                },
                ...
            ],
            "total_new_sites": int,
            "sites_with_coords": int,
            "sites_missing_coords": int,
        }
    """
    
    folder = Path(sessions_folder)
    missing = []
    sites_found = defaultdict(set)
    sites_with_coords = set()
    sites_missing_coords = set()
    
    # Scan all JSON files
    for json_file in folder.glob("**/*.json"):
        try:
            with open(json_file, "r") as f:
                inputs = json.load(f)
            
            # Find all ##new_cc_siteN## entries
            for key, value in inputs.items():
                if "##new_cc_site" in key and key.endswith("&&latitude"):
                    # Extract site name
                    site = key.replace("&&latitude", "")
                    
                    # Check if latitude is empty
                    lat_empty = (value == "" or value is None)
                    
                    # Check if longitude is empty
                    lon_key = key.replace("&&latitude", "&&longitude")
                    lon_value = inputs.get(lon_key, "")
                    lon_empty = (lon_value == "" or lon_value is None)
                    
                    # Track site
                    sites_found[site].add(json_file.name)
                    
                    if lat_empty or lon_empty:
                        sites_missing_coords.add(site)
                        
                        # Extract scenario/year from filename if possible
                        scenario = "unknown"
                        year = "unknown"
                        # e.g., "Preferred_2030_inputs.json"
                        parts = json_file.stem.split("_")
                        if len(parts) >= 2:
                            scenario = parts[0]
                            year = parts[1]
                        
                        missing.append({
                            "site": site,
                            "scenario": scenario,
                            "year": year,
                            "latitude": value,
                            "longitude": lon_value,
                            "file": str(json_file),
                        })
                    else:
                        sites_with_coords.add(site)
        
        except Exception as e:
            print(f"[ERROR] Failed to read {json_file}: {e}")
    
    return {
        "missing_coords": missing,
        "total_new_sites": len(sites_found),
        "sites_with_coords": len(sites_with_coords),
        "sites_missing_coords": len(sites_missing_coords),
        "missing_sites_list": sorted(sites_missing_coords),
    }
```

```python
result = find_new_sites_missing_coordinates("/home/307920@ontw.alfa.local/projects/epn-ma-master/src/CONNECT_CTM/logs/sessions/26_06")

print(f"Total new sites: {result['total_new_sites']}")
print(f"With coordinates: {result['sites_with_coords']}")
print(f"Missing coordinates: {result['sites_missing_coords']}")
print(f"\nMissing sites: {result['missing_sites_list']}")

# Save report
with open("missing_coordinates_report.txt", "w") as f:
    for entry in result["missing_coords"]:
        f.write(f"{entry['site']} ({entry['scenario']}/{entry['year']})\n")
        f.write(f"  Latitude: '{entry['latitude']}'\n")
        f.write(f"  Longitude: '{entry['longitude']}'\n")
        f.write(f"  File: {entry['file']}\n\n")
```
