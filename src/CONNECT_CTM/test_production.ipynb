```python

groups = {}
for row in curves_production.to_dicts():
    sector = row.get("Sector")
    cluster = row.get("Cluster")
    scenario = row.get("Scenario")
    year = str(row.get("Year"))

    key = (sector, cluster, scenario, year)
        
    if key not in groups:
        # Create a single-row DataFrame for this group
        groups[key] = pl.DataFrame(row).to_dicts()[0]
```
